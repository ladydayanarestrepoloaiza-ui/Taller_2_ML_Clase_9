# Taller abandono de producto financiero 


# 1. Estructura de los datos

```python
import pandas as pd

df = pd.read_csv("abandono_producto_financiero.csv")

df.head()
```

```python
print("Columnas:")
print(df.columns.tolist())

print("\nDimensiones:")
print(df.shape)

print("\nTipos de datos:")
print(df.dtypes)

print("\nEstadísticos descriptivos:")
display(df.describe(include="all"))
```

```python
df = df.drop(columns=["numero_fila", "id_cliente", "apellido"])

df.head()
```

```python
df["pais"] = pd.Categorical(
    df["pais"],
    categories=["Francia", "España", "Alemania"]
)

df["sexo"] = pd.Categorical(
    df["sexo"],
    categories=["Hombre", "Mujer"]
)
```

```python
df.dtypes
```

```python
df["abandono"] = df["abandono"].astype(int)

df["abandono"].value_counts()
```

```python
print("Dimensiones finales:", df.shape)
print("\nTipos de datos:")
print(df.dtypes)
```

1. ¿Cuántas variables tiene la base cruda? ¿Cuántas quedan para el modelo?
La base cruda tiene 14 variables. Después de eliminar numero_fila, id_cliente y apellido, quedan 11 variables para el modelo.
2. ¿Cuántas son cuantitativas?
Hay 8 variables cuantitativas: puntaje_crediticio, edad, antiguedad, saldo, numero_productos, tiene_tarjeta, miembro_activo y salario_estimado.
3. ¿Cuántas son categóricas o binarias (incluyendo abandono)?
Hay 3 variables categóricas o binarias: pais y sexo son categóricas, mientras que tiene_tarjeta, miembro_activo y abandono son binarias.
Corrección importante: si seguimos literalmente la clasificación del enunciado, la respuesta correcta es 5 variables categóricas o binarias, no 3. tiene_tarjeta, miembro_activo y abandono son binarias, además de pais y sexo.
4. ¿Qué implica esta mezcla para el logit en Python (C(pais), C(sexo) frente a dummies hechas a mano)?
La mezcla implica que las variables categóricas como pais y sexo no deben entrar directamente como números al modelo. En Python se pueden tratar mediante fórmulas usando C(pais) y C(sexo), lo que permite que el modelo genere las variables indicadoras automáticamente. También se podrían crear las dummies manualmente, pero habría que controlar la categoría de referencia para evitar redundancia.

```python
import matplotlib.pyplot as plt

df.boxplot(column="edad", by="abandono")

plt.title("Edad según abandono")
plt.suptitle("")
plt.xlabel("Abandono")
plt.ylabel("Edad")
plt.show()
```

```python
medianas_edad = df.groupby("abandono", observed=True)["edad"].median()

print("Mediana de edad por grupo:")
print(medianas_edad)
```

```python
from scipy.stats import chi2_contingency

tabla_activo = pd.crosstab(df["miembro_activo"], df["abandono"])

print("Tabla cruzada:")
display(tabla_activo)

chi2, p, gl, esperados = chi2_contingency(tabla_activo)

print("Chi-cuadrado:", chi2)
print("p-valor:", p)
print("Grados de libertad:", gl)
```

```python
tabla_productos = pd.crosstab(df["numero_productos"], df["abandono"])

print("Tabla cruzada:")
display(tabla_productos)

chi2, p, gl, esperados = chi2_contingency(tabla_productos)

print("Chi-cuadrado:", chi2)
print("p-valor:", p)
print("Grados de libertad:", gl)
```

### 10. ¿Qué variable parece tener mayor relación con el abandono?

Entre las variables analizadas, **numero_productos** presenta la relación más fuerte con el abandono según la prueba de ji-cuadrado. Su estadístico fue de **1503,63**, con un **p-valor reportado como 0.0**, mientras que `miembro_activo` obtuvo un chi-cuadrado de **242,99**.

Esto indica una asociación estadísticamente significativa entre el número de productos y el abandono. Los boxplots también muestran diferencias entre los grupos para edad y saldo, mientras que la diferencia en puntaje crediticio es pequeña. Por lo tanto, con base en las tablas y las pruebas de ji-cuadrado realizadas, **numero_productos es la variable que muestra la asociación más marcada con el abandono dentro de las variables analizadas**.
