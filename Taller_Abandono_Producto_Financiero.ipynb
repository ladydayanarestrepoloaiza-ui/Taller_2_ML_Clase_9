# De la recta al logit

```python
%pip install statsmodels
```

```python
import statsmodels.formula.api as smf

# Modelo OLS
modelo_ols = smf.ols("abandono ~ edad", data=df).fit()

# Modelo Logit
modelo_logit = smf.logit("abandono ~ edad", data=df).fit()

print("MODELO OLS")
print(modelo_ols.summary())

print("\nMODELO LOGIT")
print(modelo_logit.summary())
```

```python
import numpy as np

# Odds Ratio de la edad
or_edad = np.exp(modelo_logit.params["edad"])

# Probabilidad de abandono para un cliente de 40 años
prob_40 = modelo_logit.predict(pd.DataFrame({"edad": [40]}))[0]

print("Odds Ratio de edad:", or_edad)
print("Probabilidad de abandono para un cliente de 40 años:", prob_40)
print("Probabilidad en porcentaje:", prob_40 * 100)
```

11. ¿Por qué la recta de mínimos cuadrados no puede leerse como una probabilidad?
La recta OLS no puede interpretarse directamente como una probabilidad porque puede producir valores menores que 0 o mayores que 1. En este caso, aunque abandono sea una variable 0/1, OLS no restringe sus predicciones al intervalo entre 0 y 1. El modelo logit soluciona esto mediante una curva en forma de S que mantiene las probabilidades entre 0 y 1.
12. Intercepto, coeficiente de edad y Odds Ratio
Del modelo Logit obtuvimos:
Intercepto: -3.9286
Coeficiente de edad: 0.0629
Odds Ratio: 1.06497 ≈ 1.065
El intercepto del logit es -3,9286 y el coeficiente de edad es 0,0629. El Odds Ratio de la edad es 1,065.
En términos de negocio, por cada año adicional de edad, los odds de abandono aumentan aproximadamente un 6,5 %, según este modelo logit simple.
13. Probabilidad de abandono para un cliente de 40 años

Para un cliente de 40 años, el modelo logit simple estima una probabilidad de abandono de aproximadamente 19,61 %.

# 5. Varios modelos y odds ratios

```python
from numpy.random import default_rng

# Variables de modelación
X = df.drop(columns=["abandono"])
y = df["abandono"]

# Generador solicitado
rng = default_rng(8)

# Índices aleatorios para 70/30
indices = rng.permutation(len(df))
n_train = int(0.70 * len(df))

train_idx = indices[:n_train]
test_idx = indices[n_train:]

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()
y_train = y.iloc[train_idx].copy()
y_test = y.iloc[test_idx].copy()

print("Filas en train:", len(X_train))
print("Filas en test:", len(X_test))

print("\nTasa de abandono en train:", y_train.mean())
print("Tasa de abandono en test:", y_test.mean())

print("\nTasa de abandono total:", y.mean())
```

```python
# Modelos logit de menor a mayor complejidad

fit1 = smf.logit(
    "abandono ~ edad",
    data=pd.concat([X_train, y_train], axis=1)
).fit()

fit2 = smf.logit(
    "abandono ~ edad + puntaje_crediticio",
    data=pd.concat([X_train, y_train], axis=1)
).fit()

fit3 = smf.logit(
    "abandono ~ edad + puntaje_crediticio + C(sexo)",
    data=pd.concat([X_train, y_train], axis=1)
).fit()

fit4 = smf.logit(
    "abandono ~ edad + puntaje_crediticio + C(sexo) + C(pais)",
    data=pd.concat([X_train, y_train], axis=1)
).fit()

fit5 = smf.logit(
    "abandono ~ edad + puntaje_crediticio + C(sexo) + C(pais) + saldo + numero_productos + miembro_activo",
    data=pd.concat([X_train, y_train], axis=1)
).fit()

fit6 = smf.logit(
    "abandono ~ edad + puntaje_crediticio + C(sexo) + C(pais) + saldo + numero_productos + miembro_activo + antiguedad + tiene_tarjeta + salario_estimado",
    data=pd.concat([X_train, y_train], axis=1)
).fit()

# Comparación de AIC
aic = pd.DataFrame({
    "Modelo": ["fit1", "fit2", "fit3", "fit4", "fit5", "fit6"],
    "AIC": [fit1.aic, fit2.aic, fit3.aic, fit4.aic, fit5.aic, fit6.aic]
})

display(aic)
```
